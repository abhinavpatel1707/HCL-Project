```python
import pandas as pd
import numpy as np
import plotly.express as px

print("Libraries imported successfully!")
```

```python
data = {
    'Movie': [
        'Dangal', 'Jawan', 'Pathaan', '3 Idiots',
        'RRR', 'KGF 2', 'Baahubali 2', 'Pushpa 2',
        'Drishyam 2', 'Leo', 'Animal', 'Stree 2',
        'Kantara', 'Vikram', 'Brahmastra'
    ],

    'Year': [
        2016, 2023, 2023, 2009, 2022,
        2022, 2017, 2024, 2022, 2023,
        2023, 2024, 2022, 2022, 2022
    ],

    'Industry': [
        'Bollywood', 'Bollywood', 'Bollywood', 'Bollywood',
        'Telugu', 'Kannada', 'Telugu', 'Telugu',
        'Malayalam', 'Tamil', 'Bollywood', 'Bollywood',
        'Kannada', 'Tamil', 'Bollywood'
    ],

    'Genre': [
        'Drama', 'Action', 'Action', 'Comedy',
        'Action', 'Action', 'Action', 'Action',
        'Thriller', 'Action', 'Action', 'Comedy',
        'Thriller', 'Action', 'Fantasy'
    ],

    'Rating': [
        8.3, 7.0, 5.8, 8.4, 7.8,
        8.2, 8.2, 6.0, 8.2, 7.2,
        6.2, 7.0, 8.2, 8.3, 5.6
    ],

    'Collection': [
        2000, 1150, 1050, 400, 1230,
        1200, 1788, 1800, 350, 600,
        900, 875, 400, 400, 430
    ]
}

df = pd.DataFrame(data)

df.to_csv('indian_movies.csv', index=False)

print("Dataset created successfully!")
```

```python
df = pd.read_csv('indian_movies.csv')

print(df)
```

```python
print("Total Movies:", len(df))

print("Average Rating:", np.mean(df['Rating']))

print("Average Collection:", np.mean(df['Collection']))

print("Highest Collection:", np.max(df['Collection']))

print("Lowest Collection:", np.min(df['Collection']))
```

```python
print("1. Industry-wise Movies:")
print(df['Industry'].value_counts())

print("\n2. Most Popular Genre:")
print(df['Genre'].value_counts().head(1))

print("\n3. Year-wise Releases:")
print(df['Year'].value_counts().sort_index())

print("\n4. Top 5 Movies:")
print(df.nlargest(5, 'Collection')[['Movie', 'Collection']])

print("\n5. Highest Rated Movie:")
print(df.loc[df['Rating'].idxmax(), 'Movie'])

print("\n6. Average Collection:")
print(df['Collection'].mean())

print("\n7. Rating and Collection Correlation:")
print(df['Rating'].corr(df['Collection']))
```
